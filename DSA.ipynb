```python
class Student:
    def __init__(self, student_id, name, marks,):
        self.student_id = student_id
        self.name = name
        self.marks = marks
        self.next = None

    def __str__(self):
        return f"Student(id={self.student_id}, name={self.name}, marks={self.marks})"

    def __repr__(self):
        return self.__str__()
    
```

```python
class student_list:
    def __init__(self):
        self.head = None
        self.tail = None

    def add_student(self, Student):
        if self.head is None:
            self.head=Student
            self.tail=Student
        else:
            self.tail.next=Student
            self.tail=Student

    def display_students(self):
        current=self.head
        while current is not None:
            print(f"Student ID: {current.student_id}, Name: {current.name}, Marks: {current.marks}")
            current=current.next
    def search_by_id(self, id):
        current=self.head
        while current is not None:
            if current.student_id==id:
                return True
            current=current.next
        return False
    
    def delete_by_id(self, id):
        current = self.head
        previous = None

        while current is not None:
            if current.student_id == id:
                if previous is None:
                    self.head=current.next
                else:
                    previous.next=current.next
                if current == self.tail:
                    self.tail = previous
                return True
                
            previous = current
            current = current.next
        return False

    def bubble_sort_by_marks(self):
        if self.head is None:
            return

        swapped = True

        while swapped:
            swapped = False
            current = self.head

            while current.next is not None:
                if current.marks > current.next.marks:

                    current.marks, current.next.marks = (
                        current.next.marks,
                        current.marks
                    )

                    current.name, current.next.name = (
                        current.next.name,
                        current.name
                    )

                    current.student_id, current.next.student_id = (
                        current.next.student_id,
                        current.student_id
                    )

                    swapped = True

                current = current.next

    def selection_sort(self):
        if self.head is None:
            return None

        current=self.head
        while current is not None:
            minimum=current
            temp=current.next
            while temp is not None:
                if temp.marks<minimum.marks:
                    minimum=temp
                temp=temp.next

            current.marks, minimum.marks = (
            minimum.marks,
            current.marks
            )
                    
            current.name, minimum.name = (
            minimum.name,
            current.name
            )
                    
            current.student_id, minimum.student_id = (
            minimum.student_id,
            current.student_id
            )
            current=current.next

    def insertion_sort(self):
        if self.head is None:
            return None
        current=self.head.next
        while current is not None:
            key_id=current.student_id
            key_marks=current.marks
            key_name=current.name
            temp=self.head
            while temp!=current:
                if temp.marks>current.marks:
                    temp.marks, key_marks = key_marks, temp.marks
                    temp.name, key_name = key_name, temp.name
                    temp.student_id, key_id = key_id, temp.student_id

                temp=temp.next
            current.student_id=key_id
            current.marks=key_marks
            current.name=key_name

            current=current.next


        

        
        
```

```python
class stacknodes:
    def __init__(self, student):
        self.student=student
        self.next=None
```

```python
class studentstack:
    def __init__(self):
        self.top=None

    def push(self, student):
        new_node = stacknodes(student)

        new_node.next = self.top
        self.top = new_node

    def pop(self):
        if self.top is None:
            return None

        removed=self.top
        self.top=self.top.next

        return removed.student

    def peek(self):
        if self.top is None:
            return None

        return self.top.student

    def is_empty(self):
        if self.top is None:
            return True
        return False

    def delete_by_id(self, id, undo_stack):
        if self.head is None:
            return None
        current=self.head
        previous=None

        while current is not None:
            if current.student_id==id:
                undo_stack.push(current)
                if previous is None:
                    self.head=current.next
                else:
                    previous.next=current.next
                if current == self.tail:
                    self.tail = previous
                return True
                                
            previous = current
            current = current.next
        return False

    def undo_delete(self, undo_stack):
        student = undo_stack.pop()

        if student is None:
            print("Nothing to undo")
            return

        self.add_student(student)
            
        
```
